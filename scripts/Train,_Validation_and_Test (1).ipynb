```python
import pandas as pd
from google.colab import files
import io
import os

# 1. Solicitar al usuario que suba los archivos
print("Por favor, sube los 4 archivos curados desde tu ordenador:")
uploaded = files.upload()

# 2. Definir el porcentaje de partición (Ej. 80% Train+Val / 20% Test)
# Puedes ajustar este valor si en tu paper definiste otro porcentaje (ej. 0.7 o 0.85)
train_val_ratio = 0.80

for filename, content in uploaded.items():
    if filename.endswith('.csv'):
        print(f"\nProcesando: {filename}")

        # Leer el archivo CSV cargado
        df = pd.read_csv(io.BytesIO(content))

        # Asegurar el orden cronológico estricto usando la columna AÑO
        if 'AÑO' in df.columns:
            df = df.sort_values('AÑO').reset_index(drop=True)
        else:
            print(f"  Advertencia: No se encontró la columna 'AÑO' en {filename}.")

        # Calcular el índice exacto de corte
        split_index = int(len(df) * train_val_ratio)

        # División cronológica (Sin mezclar los datos)
        train_val_df = df.iloc[:split_index]
        test_df = df.iloc[split_index:]

        # Imprimir información sobre la partición para tu registro metodológico
        print(f"  - Total de años: {len(df)}")
        print(f"  - Train/Val (Años seleccionados): {len(train_val_df)} filas")
        print(f"  - Test (Años seleccionados): {len(test_df)} filas")

        # Generar nombres para los nuevos archivos
        base_name = filename.replace('.csv', '')
        train_val_filename = f"{base_name}_train_val.csv"
        test_filename = f"{base_name}_test.csv"

        # Guardar temporalmente en el entorno de Colab
        train_val_df.to_csv(train_val_filename, index=False)
        test_df.to_csv(test_filename, index=False)

        # 3. Descargar automáticamente los 2 nuevos archivos a tu ordenador
        files.download(train_val_filename)
        files.download(test_filename)

print("\n¡Proceso finalizado! Los 8 archivos se están descargando a tu ordenador.")
```
